```python
import numpy as np

# Hourly tariffs (24 hours)
tariff_rates = np.array([
    4.5, 4.5, 4.5, 4.5, 4.5, 4.5,  # 00:00 - 05:00
    7.0, 7.0, 7.0, 7.0, 7.0, 7.0,  # 06:00 - 11:00
    7.0, 7.0, 7.0, 7.0, 7.0, 7.0,  # 12:00 - 17:00
    10.0, 10.0, 10.0, 10.0,        # 18:00 - 21:00
    4.5, 4.5                       # 22:00 - 23:00
])

appliances = [
    {"name": "Refrigerator", "power": 0.2, "duration": 24, "priority": 1, "flexible": False},
    {"name": "AC / Heater", "power": 1.5, "duration": 8, "priority": 2, "flexible": True},
    {"name": "EV Scooter/Car", "power": 3.3, "duration": 4, "priority": 3, "flexible": True},
    {"name": "Washing Machine", "power": 0.5, "duration": 2, "priority": 3, "flexible": True},
]

# Track running appliances per hour
hourly_schedule = {h: [] for h in range(24)}
total_cost = 0.0

for app in appliances:
    if not app["flexible"] or app["priority"] == 1:
        run_hours = list(range(24))
    else:
        # Sort hours by cheapest tariff first
        run_hours = np.argsort(tariff_rates)[:app["duration"]]

    for hr in run_hours:
        hourly_schedule[hr].append(app["name"])
        total_cost += app["power"] * tariff_rates[hr]

print("--- APPLIANCE SCHEDULING ORDER ---")
for hr, apps in hourly_schedule.items():
    time_str = f"{hr:02d}:00 - {(hr+1):02d}:00"
    print(f"{time_str} : {', '.join(apps)}")

print("\n--- FINAL COST ---")
print(f"Total Daily Electricity Cost: ₹{total_cost:.2f}")
```
